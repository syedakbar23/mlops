```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

```python
df=pd.read_csv("/content/student-por.csv")
```

```python
df["Averaga_Grade"]=(df["G1"]+df["G2"])/2
df["Study_Efficency"]=df["studytime"]/(df["absences"]+1)
df["Parent_education"]=(df["Medu"]+df["Fedu"])/2
df ["Student_wellbeing"]=(df["famrel"]+df["freetime"]+df["health"])/3
```

```python
feature_store=df[
    [
        "Averaga_Grade",
        "Study_Efficency",
        "Parent_education",
        "Student_wellbeing"
    ]
]
feature_store.to_csv("engineering_features.csv",index=False)
print("Feature Store Created Succesfully")
```

```python
print(df.columns.tolist())
```

```python
target="G3"
X_original=df.drop(columns=["G3","Averaga_Grade","Study_Efficency","Parent_education", "Student_wellbeing"])
X_engineered=df.drop(columns=["G3"])
y=df[target]
def train_model(X,y):
  numerical_columns=X.select_dtypes(
      include=["int64","float64"]
  ).columns

  categorical_columns=X.select_dtypes(
      include=["object"]
  ).columns
  preprocessor=ColumnTransformer(
      [
          ("num",
           Pipeline([
               ("Imputer",SimpleImputer(strategy="mean"))
      ]),
           numerical_columns
           ),
          (
              "cat",
              Pipeline([
                  ("imputer",SimpleImputer(strategy="most_frequent")),
                  ("encoder",OneHotEncoder(handle_unknown="ignore"))
                  ]),
              categorical_columns)
      ]
  )
  X_train,X_test,y_train,y_test=train_test_split(
      X,
      y,
      test_size=0.20,
      random_state=42
  )
  X_train=preprocessor.fit_transform(X_train)
  X_test=preprocessor.transform(X_test)
  model=RandomForestRegressor(n_estimators=100,random_state=42)
  model.fit(X_train,y_train)
  prediction=model.predict(X_test)
  score=r2_score(y_test,prediction)
  return score
```

```python
score_original=train_model(X_original,y)
print(f"R2 Score for Original Features: {score_original}")
score_engineered=train_model(X_engineered,y)
print(f"R2 Score for Engineered Features: {score_engineered}")
```

```python
import os
import joblib
import pandas as pd
from datetime import datetime
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.metrics import mean_absolute_error
```

```python
os.makedirs("models",exist_ok=True)
```

```python
df=pd.read_csv("processed_student_data.csv")
```

```python
target="G3"
X=df.drop(columns=target)
y=df[target]
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
lr=LinearRegression()
lr.fit(X_train,y_train)
lr_prediction=lr.predict(X_test)
lr_r2=r2_score(y_test,lr_prediction)
lr_mae=mean_absolute_error(y_test,lr_prediction)
```

```python
joblib.dump(lr,"models/linear_regression_model.joblib")
```

```python
rf=RandomForestRegressor(n_estimators=100,random_state=42)
rf.fit(X_train,y_train)
rf_prediction=rf.predict(X_test)
rf_r2=r2_score(y_test,rf_prediction)
rf_mae=mean_absolute_error(y_test,rf_prediction)
```

```python
joblib.dump(rf,"models/random_forest_model.joblib")
```

```python
registry=pd.DataFrame({
    "Model":["Linear Regression","Random Forest"],
    "Version":["1.0.0","1.0.0"],
    "R2 Score":[lr_r2,rf_r2],
    "MAE":[lr_mae,rf_mae],
    "Created_date":[datetime.now(),datetime.now()]
})
best_model=registry.loc[registry["R2 Score"].idxmax(),"Model"]
print(f"Best Model: {best_model}")
registry["Status"]=registry["Model"].apply(lambda x:"Production"
if x==best_model else "Staging")
registry.to_csv("model_registry.csv",index=False)
```
